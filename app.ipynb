```python
import pandas as pd
import numpy as np
```

```python
file_name = "agriculture_data.csv"
df = pd.read_csv(file_name)
```

```python
print(df.head())
```

```python
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()
```

```python
text_columns = ["State", "Crop", "Season"]

for column in text_columns:

    # Convert to string
    df[column] = df[column].astype(str)

    # Remove extra spaces
    df[column] = df[column].str.strip()

    # Convert text to consistent format
    df[column] = df[column].str.title()

print("\nText columns cleaned.")
```

```python
df["Area"] = pd.to_numeric(df["Area"], errors="coerce")
df["Production"] = pd.to_numeric(df["Production"], errors="coerce")

print("Numerical columns converted.")
```

```python
df["Area"] = df["Area"].fillna(df["Area"].median())
df["Production"] = df["Production"].fillna(df["Production"].median())
```

```python
df["Date"] = pd.to_datetime(df["Date"], errors="coerce")

# Fill missing dates using the most common date
if df["Date"].notna().any():
    common_date = df["Date"].mode()[0]
    df["Date"] = df["Date"].fillna(common_date)

print("Date column converted.")
```

```python
df["yield"] = np.where(
    df["Area"] > 0,
    df["Production"] / df["Area"],
    0
)

print("\nYield column created.")
```

```python

df = df[df["Area"] >= 0]
df = df[df["Production"] >= 0]
```

```python
print(df.isnull().sum())
```

```python
output_file = "agriculture_cleaned.csv"

df.to_csv(output_file, index=False)

print("\n" + "=" * 60)
print(" CLEANING COMPLETED SUCCESSFULLY!")
print("=" * 60)

print("\nCleaned dataset saved as:")
print(output_file)

print("\nTotal rows after cleaning:", len(df))
print("Total columns after cleaning:", len(df.columns))
```
